# AUT – Alternative Uses Task

Generate a creative-uses prompt for a common object.

**Usage**
1. Call `instruct.instruct(object=None, seed=None)`.
   - If `object` is omitted a random object is sampled from the standard list.
2. The returned `response_format` already contains the original `object` so later evaluation has full context.
3. Collect a list of short creative-use phrases.

**Note:** `evaluate` is still a placeholder – scoring (fluency / originality / flexibility) will be added later.

````python
import instruct, evaluate

# Random object
stim = instruct.instruct(seed=0)
for k, v in stim.items():
    print(k)
    print(v)
    print()

# Fixed object
stim_brick = instruct.instruct(object="brick")
print("object =", stim_brick["object"])
````

### Expected response shape

```python
{
  "object": "brick",
  "uses": ["doorstop", "paperweight", "makeshift hammer", ...]
}
```

The original object travels with the answers so evaluation can judge appropriateness without external lookup.
