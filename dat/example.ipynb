# DAT – Divergent Association Task

Generate instructions for producing maximally different nouns, then score the list with average pairwise cosine distance.

**Usage**
1. Call `instruct.instruct(n_words=10)` to obtain the prompt and expected response format.
2. Collect a list (or dict) of words that follows the format.
3. Pass the words to `evaluate.evaluate(...)` to get the DAT score (higher = more divergent).

The classic Olson scoring uses the first 7 unique valid words.

```python
import instruct, evaluate

stim = instruct.instruct(n_words=10)
for k, v in stim.items():
    print(k)
    print(v)
    print()
```
