# DAT – Divergent Association Task

Generate 10 nouns that are as different from each other as possible.
Higher average pairwise distance (via GloVe) = higher creativity / divergence score.

## How to use

1. Call `instruct.instruct()` to get the prompt and expected response format.
2. Give the instructions to a human or AI and collect the answers.
3. Pass the answers to `evaluate.evaluate(...)` to obtain a score.

Run this notebook from inside the `dat/` folder (or adjust the import path).

### 1. Generate instructions

```python
import instruct, evaluate

stim = instruct.instruct(n_words=10)

print("Test:", stim["test"])
print()
print(stim["instructions"])
print()
print("Expected response format:")
print(stim["response_format"])
```

### 2. Score a response

Pass either a list of words or a dict matching the response format.
The score is the average pairwise cosine *distance* of the first 7 valid unique words × 100 (higher = more divergent).

```python
# Example answers (replace with real responses)
answers = ["cat", "thimble", "hippo", "violin", "machinery", "prickle", "cloud"]

result = evaluate.evaluate(answers, model_key="glove-6b-300d")
print(result)
```
