# CAT – Convergent Association Task

Generate word-pair stimuli and (optionally) score a response with GloVe proximity.

**Usage**
1. Call `instruct.instruct(...)` to obtain instructions + items.
2. Collect a response that follows the `response_format`.
3. Pass the response dict to `evaluate.evaluate(...)` to get a proximity score.

Parameters for `instruct`:
- `single_item=False` – return multiple pairs (default) or a single pair.
- `n_words=10` – number of pairs when `single_item=False`.
- `seed=None` – optional random seed for reproducibility.

```python
import instruct, evaluate

# Multi-item stimulus
stim = instruct.instruct(n_words=5, seed=42)
for k, v in stim.items():
    print(k)
    print(v)
    print()
```
