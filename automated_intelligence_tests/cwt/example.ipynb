# CWT – Creative Writing Task

Generate a short-story prompt that must include one, two or three cue words.

**Usage**
1. Call `instruct.instruct(cue=None, n_words=3, seed=None)`.
   - If `cue` is omitted a random set of length `n_words` is sampled from the standard sets (1, 2 or 3).
2. The returned `response_format` already contains the original `cue` so later evaluation has full context.
3. Collect a 5–6 sentence creative story that uses every cue word.

**Note:** `evaluate` is still a placeholder – creativity scoring (e.g. DSI) will be added later.

**Source:** Johnson et al. (2023), Behavior Research Methods. DOI: 10.3758/s13428-022-01986-2

````python
import instruct, evaluate

# Random 3-word cues (default)
stim = instruct.instruct(seed=0)
print("=== n_words=3 (default) ===")
for k, v in stim.items():
    print(k, ":", v)
print()

# 1-word
stim = instruct.instruct(n_words=1, seed=0)
print("=== n_words=1 ===")
for k, v in stim.items():
    print(k, ":", v)
print()

# 2-word
stim = instruct.instruct(n_words=2, seed=0)
print("=== n_words=2 ===")
for k, v in stim.items():
    print(k, ":", v)
print()

# Fixed cue list
stim = instruct.instruct(cue=["moon", "river", "whisper"])
print("=== fixed cue ===")
for k, v in stim.items():
    print(k, ":", v)
print()
````

### Expected response shape

```python
{
  "cue": ["moon", "river", "whisper"],
  "story": "Five to six sentences that include every cue word …"
}
```

The original cue travels with the story so evaluation can verify that every cue word appears and judge appropriateness without external lookup.
