# WRT – Creative Writing Task

Generate a short-story prompt that must include three cue words.

**Usage**
1. Call `instruct.instruct(cues=None, seed=None)`.
   - If `cues` is omitted a random triple is sampled from the standard set.
2. The returned `response_format` already contains the original `cues` so later evaluation has full context.
3. Collect a 5–6 sentence creative story that uses every cue word.

**Note:** `evaluate` is still a placeholder – creativity scoring will be added later.

````python
import instruct, evaluate

# Random cue triple
stim = instruct.instruct(seed=0)
for k, v in stim.items():
    print(k)
    print(v)
    print()

# Fixed cues
stim_fixed = instruct.instruct(cues=["moon", "river", "whisper"])
print("cues =", stim_fixed["cues"])
````

### Expected response shape

```python
{
  "cues": ["moon", "river", "whisper"],
  "story": "Five to six sentences that include every cue word …"
}
```

The original cues travel with the story so evaluation can verify that every cue appears and judge appropriateness without external lookup.
